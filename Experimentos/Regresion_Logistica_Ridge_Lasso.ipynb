```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.preprocessing import LabelEncoder, StandardScaler
import pandas as pd
import numpy as np
```

## Generación del modelo

```python
def train_model_with_hyperparameters_ridge_lasso():
    # Cargar datos
    df_train = pd.read_csv('../Data/train.csv')
    df_val = pd.read_csv('../Data/validation.csv')
    df_test = pd.read_csv('../Data/test.csv')

    # Codificación de categóricas
    df_combined = pd.concat([df_train, df_val])
    label_encoders = {}
    for col in df_combined.select_dtypes(include='object').columns:
        le = LabelEncoder()
        df_combined[col] = le.fit_transform(df_combined[col].astype(str))
        label_encoders[col] = le

    df_train = df_combined.iloc[:len(df_train)].copy()
    df_val = df_combined.iloc[len(df_train):].copy()

    for col in label_encoders:
        df_test[col] = label_encoders[col].transform(df_test[col].astype(str))

    # Definir X e y
    X_train = df_train.drop(columns=['G1', 'G2', 'G3'])
    y_train = df_train['G3']
    X_val = df_val.drop(columns=['G1', 'G2', 'G3'])
    y_val = df_val['G3']

    # Hiperparámetros
    alphas = [0.1, 1.0]
    fit_intercepts = [True, False]
    normalizes = [True, False]
    model_types = ['Ridge', 'Lasso']

    for model_type in model_types:
        for alpha in alphas:
            for fi in fit_intercepts:
                for norm in normalizes:
                    r2, mae, mse, rmse = generate_model_ridge_lasso(X_train, X_val, y_train, y_val, alpha, fi, norm, model_type)
                    print(f"{model_type} | alpha={alpha} | fit_intercept={fi} | normalize={norm}")
                    print(f"R²: {r2:.4f} | MAE: {mae:.4f} | MSE: {mse:.4f} | RMSE: {rmse:.4f}")
                    print('-' * 80)
```
